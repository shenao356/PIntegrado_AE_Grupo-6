### Paso 1: Importación de librerías y configuración inicial
Para este ejercicio utilizamos `pandas` para manipular tablas de datos, `datasets` de Hugging Face para descargar la información directamente desde su repositorio, y la librería estándar `pathlib` y `os` para gestionar las rutas de los archivos de manera relativa y ordenada.

```python
import os
import sys
from pathlib import Path
import pandas as pd
from datasets import load_dataset

# Opciones de visualización para que pandas muestre la tabla de forma cómoda
pd.set_option('display.max_columns', 20)
pd.set_option('display.width', 1000)

print("Librerías importadas correctamente.")
print(f"Versión de Pandas: {pd.__version__}")
```

### Paso 2: Carga del dataset desde Hugging Face
El dataset seleccionado para nuestro proyecto integrador es:
- **Identificador en Hugging Face:** `MongoDB/whatscooking.restaurants`
- **Split utilizado:** `train`
- **Tema:** Sector gastronómico (restaurantes de la ciudad de Nueva York).

Este dataset es el mismo con el que trabajamos en la actividad anterior, lo cual nos permite mantener la coherencia y continuidad en todo el proyecto.

```python
DATASET_ID = "MongoDB/whatscooking.restaurants"
SPLIT = "train"

print(f"Descargando y cargando el dataset '{DATASET_ID}' (split='{SPLIT}')...")

# Descarga directa desde Hugging Face y conversión a pandas DataFrame
dataset = load_dataset(DATASET_ID, split=SPLIT)
df = dataset.to_pandas()

print("¡Dataset cargado con éxito en el DataFrame de Pandas!")
print(f"Total de registros cargados: {len(df):,} filas")
```

### Paso 4: Exploración de columnas y tipos de datos

### Paso 5: Verificación de valores faltantes (nulos) por columna
Un paso fundamental en la analítica de datos es saber cuántos datos faltan en cada columna (`NaN` o `None`). Esto nos permite saber con qué variables podemos trabajar directamente y cuáles requerirán un tratamiento especial o limpieza en la siguiente etapa.

### Paso 6: Comprensión del contenido y relación con la necesidad

#### 1. ¿Qué representa cada fila?
Cada fila (registro) de este DataFrame representa **un restaurante físico o establecimiento gastronómico individualmente registrado** en la ciudad de Nueva York. No se trata de un ticket de compra ni de una factura, sino de la ficha técnica y comercial del negocio, identificada con su `restaurant_id` único, su nombre comercial, su ubicación geográfica, su oferta culinaria y la valoración de sus clientes.

#### 2. ¿Cómo respaldan las variables la necesidad de nuestro proyecto?
Nuestra necesidad de negocio busca entender cómo se distribuye la oferta restaurantera en Nueva York, qué zonas tienen mayor concentración y qué tipos de comida son los más comunes y mejor valorados. Las variables respaldan este propósito de la siguiente manera:
- **`borough` (Barrio o Distrito):** Permite agrupar los restaurantes por zonas (Manhattan, Brooklyn, Queens, Bronx, Staten Island) para analizar la cobertura y saturación del mercado.
- **`cuisine` (Tipo de Cocina):** Identifica la especialidad culinaria (Americana, China, Italiana, Mexicana, etc.), clave para entender qué oferta predomina.
- **`stars` (Calificación en estrellas de 1.0 a 5.0):** Permite evaluar el nivel de satisfacción que tienen los clientes con respecto a cada restaurante o tipo de cocina.
- **`review_count` (Cantidad de reseñas):** Representa el volumen de interacción y popularidad del establecimiento en la plataforma.
- **`PriceRange` (Rango de precios de 1 a 4):** Indica el nivel socioeconómico de la oferta (económico, moderado, costoso, muy costoso).
- **`OutdoorSeating` y `TakeOut`:** Describen facilidades y servicios que ofrecen comodidades adicionales a los comensales.

Por todo esto, confirmamos que el dataset es completamente pertinente para nuestro objetivo descriptivo.

### Paso 7: Generación automática del archivo de resumen `.txt`
La guía de la actividad solicita que el notebook genere automáticamente un archivo `.txt` que resuma el DataFrame analizado, guardándolo en la carpeta `resultados/` con codificación UTF-8.

El archivo debe incluir:
1. Identificador del dataset y split utilizado.
2. Cantidad de filas y columnas.
3. Nombres y tipos de las columnas.
4. Cantidad de valores nulos por columna.
5. Estadísticas descriptivas básicas.

```python
# Determinamos la ruta de salida relativa para que funcione perfectamente
# tanto si se ejecuta desde la raíz del proyecto como desde la subcarpeta notebooks/
current_dir = Path.cwd()

if current_dir.name == "notebooks":
    carpeta_resultados = (current_dir.parent / "resultados").resolve()
elif (current_dir / "AE2" / "resultados").exists() or (current_dir / "AE2").exists():
    carpeta_resultados = (current_dir / "AE2" / "resultados").resolve()
else:
    carpeta_resultados = (current_dir / "resultados").resolve()

carpeta_resultados.mkdir(parents=True, exist_ok=True)
ruta_archivo_txt = carpeta_resultados / "resumen_dataframe.txt"

print(f"Ruta de destino del resumen: {ruta_archivo_txt}")

# Redactamos el contenido del resumen en texto plano
lineas = []
lineas.append("=" * 70)
lineas.append("REPORTE DE RESUMEN DEL DATAFRAME - PROYECTO INTEGRADOR (AE2)")
lineas.append("=" * 70)
lineas.append(f"Identificador del Dataset en Hugging Face: {DATASET_ID}")
lineas.append(f"Split utilizado: {SPLIT}")
lineas.append(f"Muestra o total: Conjunto completo cargado directamente desde el split oficial.")
lineas.append(f"Total de Filas (Observaciones): {len(df):,}")
lineas.append(f"Total de Columnas (Variables): {len(df.columns)}")
lineas.append("-" * 70)
lineas.append("LISTADO DE COLUMNAS Y TIPOS DE DATOS:")
lineas.append("-" * 70)
for col, dtype in zip(df.columns, df.dtypes):
    lineas.append(f" - {col:<20} | Tipo: {str(dtype):<15}")

lineas.append("-" * 70)
lineas.append("CONTEO DE VALORES NULOS POR COLUMNA:")
lineas.append("-" * 70)
for col in df.columns:
    nulos = int(df[col].isna().sum())
    pct = (nulos / len(df)) * 100
    lineas.append(f" - {col:<20} | Nulos: {nulos:>6} ({pct:>6.2f} %)")

lineas.append("-" * 70)
lineas.append("ESTADÍSTICAS BÁSICAS DE VARIABLES NUMÉRICAS CLAVE:")
lineas.append("-" * 70)
cols_numericas = ['stars', 'review_count', 'PriceRange']
for col in cols_numericas:
    if col in df.columns:
        s = df[col].dropna()
        lineas.append(f"Variable: {col}")
        lineas.append(f"   Conteo válidos: {len(s):,}")
        lineas.append(f"   Promedio:       {s.mean():.2f}")
        lineas.append(f"   Desv. Estándar: {s.std():.2f}")
        lineas.append(f"   Mínimo:         {s.min():.2f}")
        lineas.append(f"   Mediana (50%):  {s.median():.2f}")
        lineas.append(f"   Máximo:         {s.max():.2f}")
        lineas.append("")

lineas.append("=" * 70)
lineas.append("Fin del reporte automático generado por el notebook.")
lineas.append("=" * 70)

# Escribimos el archivo asegurando codificación UTF-8
contenido_txt = "\n".join(lineas)
with open(ruta_archivo_txt, "w", encoding="utf-8") as f:
    f.write(contenido_txt)

# También guardamos una copia en resultados de la raíz si se ejecuta desde raíz para máxima compatibilidad
if (current_dir / "resultados").resolve() != carpeta_resultados:
    carpeta_raiz_res = current_dir / "resultados"
    carpeta_raiz_res.mkdir(parents=True, exist_ok=True)
    with open(carpeta_raiz_res / "resumen_dataframe.txt", "w", encoding="utf-8") as f:
        f.write(contenido_txt)

print("¡Archivo 'resumen_dataframe.txt' generado con éxito!")
```

### Paso 8: Verificación automática de la lectura del archivo `.txt`
Para cumplir estrictamente con la rúbrica ("verifica su contenido y presenta resultados visibles"), abrimos el archivo recién guardado, comprobamos su tamaño en disco y mostramos su contenido en pantalla.

```python
if ruta_archivo_txt.exists():
    tamano_bytes = ruta_archivo_txt.stat().st_size
    print(f"Estado: Archivo encontrado exitosamente.")
    print(f"Tamaño del archivo: {tamano_bytes:,} bytes.\n")
    print("--- INICIO DEL CONTENIDO LEÍDO DESDE EL ARCHIVO TXT ---")
    with open(ruta_archivo_txt, "r", encoding="utf-8") as f:
        print(f.read())
    print("--- FIN DEL CONTENIDO LEÍDO DESDE EL ARCHIVO TXT ---")
else:
    print("ERROR: No se encontró el archivo generado en la ruta esperada.")
```
